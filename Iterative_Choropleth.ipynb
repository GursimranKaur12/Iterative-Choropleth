```python
import pandas as pd
import plotly.express as px
from datetime import datetime
import pytz
```

```python
df_play_store = pd.read_csv('Play Store Data.csv')
```

```python
df_ev_data = pd.read_csv('Global EV Data 2024.csv')
```

```python
def clean_installs(val):
  if isinstance(val, str):
    if val == 'Free':
      return 0
    else:
      return int(val.replace('+', '').replace(',', ''))
  else:
    return 0
```

```python
df_play_store['Installs'] = df_play_store['Installs'].apply(clean_installs)
```

```python
filtered_df = df_play_store[~df_play_store['Category'].str.startswith(('A', 'C', 'G', 'S'), na=False)]
```

```python
top_categories = filtered_df.groupby('Category')['Installs'].sum().nlargest(5).index
final_df_play_store = filtered_df[filtered_df['Category'].isin(top_categories)]
```

```python
ist = pytz.timezone('Asia/Kolkata')
current_time= datetime.now(ist).time()
```

```python
if 18 <= current_time.hour <20:
  fig = px.choropleth(final_df_play_store, locations=['Global']*len(final_df_play_store),
                      locationmode='country names',
                      color='Category',
                      hover_name='Category',
                      title = 'Top 5 Categories Install Distribution',
                      color_continuous_scale=px.colors.sequential.Plasma)
  fig.update_geos(visible=False)
  fig.show()
else:
  print("Visualization only available between 6 PM - 8 PM IST")
```
